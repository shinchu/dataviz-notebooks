## 整然データにする

整然データは、次の4つの条件を満たすデータなのでした。

1. 1つの列は、1つの変数を表す。
2. 1つの行は、1つの観測を表す。
3. 1つのセル（特定の列の特定の行）は、1つの値を表す。
4. 1つの表は、1つの観測単位を持つ（異なる観察単位が混ざっていない）。

今扱っているデータを整然データにするには、3つの列を作ればよさそうです。

1. `Date`: 日付
2. `Prefecture`: 都道府県
3. `Cases`: 感染者数

列が可変で一行に複数の観測があるデータを「横持ちのデータ」とも言います。この場合は、横持ちから縦持ちへの変換という操作を行います。

`Date`はすでに列になっていて、これは動かしたくないので、まずはこれをindexに設定します。

```python
df = df.set_index("Date")
```

```python
df
```

次に、`stack`メソッドを使って現在の列を一列に変換します。

```python
df = df.stack()
```

```python
df
```

最後に、`Date`を普通の列に戻してあげます。

```python
df = df.reset_index()
```

```python
df
```

列名を変更します。

```python
df.columns = ["Date", "Prefecture", "Cases"]
```

```python
df
```

これで横持ちのデータを縦持ちに変換できました。

ただ、整然データにはもう一歩足りないようです。

都道府県ごとの観測値の中に観測値の合計（`ALL`）が混ざっているので、4つ目の条件（1つの表は、1つの観測単位を持つ）を満たしていないのですね。

```python
len(df.groupby("Prefecture"))
```

都道府県でグルーピングすると48あることが分かります。

というわけで、`ALL`の値を削除します。

縦持ちデータのまま`ALL`を削除することもできますし、

```python
df[df["Prefecture"] != "ALL"]
```

```python
df.query("Prefecture != 'ALL'")
```

横持ちデータに戻してから列ごとALLを削除することもできます。縦持ちから横持ちへの変換は`pivot`を使います。

```python
df.pivot(index="Date", columns="Prefecture", values="Cases").drop(columns="ALL")
```

```python
df = df.pivot(index="Date", columns="Prefecture", values="Cases").drop(columns="ALL").stack().reset_index()
```

```python
df.columns = ["Date", "Prefecture", "Cases"]
```

```python
df
```

```python
len(df.groupby("Prefecture"))
```

これで、47都道府県になりました。

次に、データをより使いやすくするために日付を適切なデータ型にします。

```python
df["Date"].apply(type)
```

現在は、文字列型だということが分かります。

これを日付型に変換します。ある程度整っているデータであれば、pandasの`to_datetime`関数を使うことができます。

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df
```

```python
df["Date"].apply(type)
```

日付型になりました。

ここから年月日を取り出して、それぞれ別の変数として格納します。

```python
df["Year"] = pd.DatetimeIndex(df["Date"]).year
df["Month"] = pd.DatetimeIndex(df["Date"]).month
df["Day"] = pd.DatetimeIndex(df["Date"]).day
```

```python
df
```

これで、整然データを作ることができました。

## データを使ってみる

データを使って簡単な可視化を行います。

データ可視化ライブラリのAltairについては来週詳しく学びますので、今はコードの意味が分からなくても大丈夫です。

```python
# Altairのインストール
!pip install --upgrade "altair>=6"
```

```python
# Altairのインポート
import altair as alt
```

```python
# Altairの読み込み行数制限を無効化する（デフォルトでは500行しか読み込めない）

alt.data_transformers.disable_max_rows()
```

```python
chart = alt.Chart(df)
```

```python
chart.mark_line().encode(
    x="Date",
    y="sum(Cases)",
    tooltip=["Date", "sum(Cases)"]
).interactive()
```

時系列でみる総感染者数。Googleなどで見る図と同じ形ですね。

```python
chart.mark_point().encode(
    x="Month",
    y="mean(Cases)",
    color="Prefecture",
    tooltip=["Year", "Prefecture", "mean(Cases)"]
).interactive()
```

月ごとの平均感染者数はこんな感じ。冬のほうがウイルスが活発になるはずですが、今年の夏に爆発的に増えたので、その影響を受けているようです。

```python
topten = df.groupby('Prefecture').sum()['Cases'].sort_values(ascending=False)[:10].keys()
```

```python
topten
```

```python
df_topten = df[df['Prefecture'].isin(topten)]
```

```python
df_topten
```

```python
chart = alt.Chart(df_topten)
```

```python
chart.mark_bar().encode(
    x='Year:O',
    y='sum(Cases)',
    color='Month:O',
    column='Prefecture',
    tooltip=['Month', 'sum(Cases)']
)
```

感染者数が最も多い10都道府県の年ごとの総感染者数はこんな感じです。
